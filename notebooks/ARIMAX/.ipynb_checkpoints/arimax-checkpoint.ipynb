# ARIMAX: СКР с экзогенными переменными

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
warnings.filterwarnings("ignore")

# ============================================================
# ФАЙЛЫ ДАННЫХ
# ============================================================
TRAIN_FILE = "train_СКР.xlsx"
ACTUAL_FILE = "СКР_2024.xlsx"
EXOG_FILE = "exog_data.xlsx"  # Файл с экзогенными переменными

# ============================================================
# НАСТРОЙКИ ЭКЗОГЕННЫХ ПЕРЕМЕННЫХ
# ============================================================
# Укажите названия столбцов с экзогенными переменными
# Например: ['ВРП', 'Безработица', 'Средняя_зарплата', 'Инфляция']
EXOG_COLUMNS = ['ВРП', 'Безработица']  # <-- ИЗМЕНИТЕ ПОД ВАШИ ДАННЫЕ


# Функция очистки чисел с запятых и пробелов
def clean_numeric(series):
    return (
        series.astype(str)
        .str.replace('\xa0', '', regex=False)
        .str.replace(' ', '')
        .str.replace(',', '.')
        .astype(float)
    )


# Функция для стандартизации названий регионов
def standardize_region_names(df):
    """Стандартизация названий регионов"""
    df_clean = df.copy()

    replacements = {
        'Ненецкий авт.округ': 'Ненецкий автономный округ',
        'Hенецкий авт.округ': 'Ненецкий автономный округ',
        '  Ненецкий автономный округ': 'Ненецкий автономный округ',
        'Ямало-Ненецкий авт.округ': 'Ямало-Ненецкий автономный округ',
        'Ямало-Hенецкий авт.округ': 'Ямало-Ненецкий автономный округ',
        '  Ямало-Ненецкий автономный округ': 'Ямало-Ненецкий автономный округ',
        'Ханты-Мансийский авт.округ-Югра': 'Ханты-Мансийский автономный округ - Югра',
        '  Ханты-Мансийский автономный округ - Югра': 'Ханты-Мансийский автономный округ - Югра',
        'Республика Татарстан(Татарстан)': 'Республика Татарстан',
        'Чувашская Республика(Чувашия)': 'Чувашская Республика',
        'Республика Северная Осетия- Алания': 'Республика Северная Осетия-Алания',
        'Oмская область': 'Омская область',
        'Hижегородская область': 'Нижегородская область',
        'г. Севастополь': 'г.Севастополь',
        'г.Москва': 'г.Москва',
        'г.Санкт-Петербург': 'г.Санкт-Петербург',
        'Чукотский авт.округ': 'Чукотский автономный округ',
    }

    df_clean['Регион'] = df_clean['Регион'].replace(replacements)
    df_clean['Регион'] = df_clean['Регион'].str.strip()
    return df_clean


def prepare_exog_data(df_exog, region, years, exog_cols):
    """
    Подготовка экзогенных переменных для региона.
    Возвращает DataFrame с экзогенными переменными для указанных лет.
    """
    region_data = df_exog[df_exog['Регион'] == region].copy()
    region_data = region_data[region_data['Год'].isin(years)].sort_values('Год')
    
    if len(region_data) != len(years):
        return None
    
    # Очистка числовых данных
    for col in exog_cols:
        if col in region_data.columns:
            region_data[col] = clean_numeric(region_data[col])
    
    return region_data[exog_cols].values


def fit_arimax(endog, exog_train, exog_forecast, order_range=(3, 2, 3)):
    """
    Подбор лучшей ARIMAX модели по AIC.
    
    Parameters:
    -----------
    endog : array-like
        Эндогенная переменная (целевой ряд)
    exog_train : array-like
        Экзогенные переменные для обучения
    exog_forecast : array-like
        Экзогенные переменные для прогноза (на 1 шаг вперёд)
    order_range : tuple
        Диапазоны для (p, d, q)
    
    Returns:
    --------
    best_forecast : float
        Лучший прогноз
    best_aic : float
        AIC лучшей модели
    best_order : tuple
        Параметры лучшей модели (p, d, q)
    """
    best_forecast = np.nan
    best_aic = np.inf
    best_order = None
    
    p_max, d_max, q_max = order_range
    
    for p in range(0, p_max):
        for d in range(0, d_max):
            for q in range(0, q_max):
                try:
                    # ARIMAX: ARIMA с экзогенными переменными
                    model = ARIMA(
                        endog,
                        exog=exog_train,
                        order=(p, d, q)
                    )
                    fitted = model.fit()
                    
                    # Прогноз с экзогенными переменными для прогнозного периода
                    forecast = fitted.forecast(steps=1, exog=exog_forecast)[0]
                    
                    if fitted.aic < best_aic and np.isfinite(forecast):
                        best_aic = fitted.aic
                        best_forecast = forecast
                        best_order = (p, d, q)
                except:
                    continue
    
    return best_forecast, best_aic, best_order


# ============================================================
# ЗАГРУЗКА ДАННЫХ
# ============================================================
print("Загрузка данных...")
df_train = pd.read_excel(TRAIN_FILE)
df_actual = pd.read_excel(ACTUAL_FILE)

# Загрузка экзогенных данных
try:
    df_exog = pd.read_excel(EXOG_FILE)
    df_exog = standardize_region_names(df_exog)
    df_exog['Год'] = pd.to_numeric(df_exog['Год'], errors='coerce')
    USE_EXOG = True
    print(f"Экзогенные переменные загружены: {EXOG_COLUMNS}")
except FileNotFoundError:
    print(f"ВНИМАНИЕ: Файл {EXOG_FILE} не найден. Модель будет работать как обычная ARIMA.")
    USE_EXOG = False
    df_exog = None

# Применяем стандартизацию
print("Стандартизация названий регионов...")
df_train = standardize_region_names(df_train)
df_actual = standardize_region_names(df_actual)

# Очистка СКР
df_train['СКР'] = clean_numeric(df_train['СКР'])
df_actual['СКР'] = clean_numeric(df_actual['СКР'])

# Убедимся, что годы — целые числа
df_train['Год'] = pd.to_numeric(df_train['Год'], errors='coerce')
df_actual['Год'] = pd.to_numeric(df_actual['Год'], errors='coerce')

# Берём только 2014–2023
df_train = df_train[(df_train['Год'] >= 2014) & (df_train['Год'] <= 2023)]

# Создаём словарь: регион - фактический СКР за 2024
actual_dict = dict(zip(df_actual['Регион'], df_actual['СКР']))

# Общие регионы
common_regions = set(df_train['Регион']) & set(df_actual['Регион'])
print(f"Общих регионов между датасетами: {len(common_regions)}")


# ============================================================
# ПРОГНОЗИРОВАНИЕ ARIMAX
# ============================================================
results = []

for region in sorted(common_regions):
    ts_data = df_train[df_train['Регион'] == region].sort_values('Год')
    if len(ts_data) < 10:
        print(f"Пропущен {region}: мало данных ({len(ts_data)} лет)")
        continue

    scr_values = ts_data['СКР'].values
    years = ts_data['Год'].values
    actual_2024 = actual_dict[region]

    # Подготовка экзогенных данных
    exog_train = None
    exog_forecast = None
    
    if USE_EXOG and df_exog is not None:
        # Экзогенные для обучения (2014-2023)
        exog_train = prepare_exog_data(df_exog, region, years, EXOG_COLUMNS)
        
        # Экзогенные для прогноза (2024)
        exog_forecast = prepare_exog_data(df_exog, region, [2024], EXOG_COLUMNS)
        
        if exog_train is None or exog_forecast is None:
            print(f"ВНИМАНИЕ: Нет экзогенных данных для {region}, используется ARIMA")
            exog_train = None
            exog_forecast = None

    # Подбор лучшей модели
    best_forecast, best_aic, best_order = fit_arimax(
        scr_values, 
        exog_train, 
        exog_forecast
    )

    # Fallback на ARIMA(1,1,1)
    if pd.isna(best_forecast):
        try:
            fallback = ARIMA(scr_values, exog=exog_train, order=(1,1,1)).fit()
            best_forecast = fallback.forecast(steps=1, exog=exog_forecast)[0]
            best_order = (1, 1, 1)
        except:
            # Fallback без экзогенных
            try:
                fallback = ARIMA(scr_values, order=(1,1,1)).fit()
                best_forecast = fallback.forecast(steps=1)[0]
                best_order = (1, 1, 1)
            except:
                print(f"Не удалось построить модель для {region}")
                continue

    # Метрики
    abs_error = abs(best_forecast - actual_2024)
    rmse = np.sqrt(mean_squared_error([actual_2024], [best_forecast]))
    mae = mean_absolute_error([actual_2024], [best_forecast])

    model_type = "ARIMAX" if (exog_train is not None) else "ARIMA"
    
    results.append({
        'Регион': region,
        'Модель': model_type,
        'Порядок_(p,d,q)': str(best_order),
        'Прогноз_СКР_2024': round(best_forecast, 4),
        'Факт_СКР_2024': round(actual_2024, 4),
        'Абсолютная_ошибка': round(abs_error, 4),
        'RMSE': round(rmse, 4),
        'MAE': round(mae, 4)
    })

    print(f"{region} [{model_type}{best_order}]: прогноз = {best_forecast:.4f}, факт = {actual_2024:.4f}, ошибка = {abs_error:.4f}")


# ============================================================
# ИТОГИ
# ============================================================
if results:
    results_df = pd.DataFrame(results)
    overall_rmse = np.sqrt(mean_squared_error(results_df['Факт_СКР_2024'], results_df['Прогноз_СКР_2024']))
    overall_mae = mean_absolute_error(results_df['Факт_СКР_2024'], results_df['Прогноз_СКР_2024'])

    print("\n" + "="*60)
    print(f"Общий RMSE по СКР: {overall_rmse:.4f}")
    print(f"Общий MAE по СКР: {overall_mae:.4f}")
    print(f"Использованы экзогенные переменные: {USE_EXOG}")
    if USE_EXOG:
        print(f"Переменные: {EXOG_COLUMNS}")
    print("="*60)

    # Сохраняем результат
    output_file = "прогноз_СКР_2024_ARIMAX_результаты.xlsx"
    results_df.to_excel(output_file, index=False)
    print(f"\nРезультаты сохранены в {output_file}")
```

# ARIMAX: ОПЖ с экзогенными переменными

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
warnings.filterwarnings("ignore")

# ============================================================
# ФАЙЛЫ ДАННЫХ
# ============================================================
FILE_PATH = "train_ОПЖ.xlsx"
EXOG_FILE = "exog_data.xlsx"  # Файл с экзогенными переменными

# ============================================================
# НАСТРОЙКИ ЭКЗОГЕННЫХ ПЕРЕМЕННЫХ
# ============================================================
# Укажите названия столбцов с экзогенными переменными
# Для ОПЖ могут быть релевантны: расходы на здравоохранение, 
# уровень доходов, экология, доступность медицины и т.д.
EXOG_COLUMNS = ['Расходы_на_здравоохранение', 'Доходы_населения']  # <-- ИЗМЕНИТЕ ПОД ВАШИ ДАННЫЕ


# Функция очистки чисел
def clean_numeric(series):
    return (
        series.astype(str)
        .str.replace('\xa0', '', regex=False)
        .str.replace(' ', '')
        .str.replace(',', '.')
        .astype(float)
    )


# Функция стандартизации регионов
def standardize_region_names(df):
    df_clean = df.copy()
    replacements = {
        'Ненецкий авт.округ': 'Ненецкий автономный округ',
        'Hенецкий авт.округ': 'Ненецкий автономный округ',
        '  Ненецкий автономный округ': 'Ненецкий автономный округ',
        'Ямало-Ненецкий авт.округ': 'Ямало-Ненецкий автономный округ',
        'Ямало-Hенецкий авт.округ': 'Ямало-Ненецкий автономный округ',
        '  Ямало-Ненецкий автономный округ': 'Ямало-Ненецкий автономный округ',
        'Ханты-Мансийский авт.округ-Югра': 'Ханты-Мансийский автономный округ - Югра',
        '  Ханты-Мансийский автономный округ - Югра': 'Ханты-Мансийский автономный округ - Югра',
        'Республика Татарстан(Татарстан)': 'Республика Татарстан',
        'Чувашская Республика(Чувашия)': 'Чувашская Республика',
        'Республика Северная Осетия- Алания': 'Республика Северная Осетия-Алания',
        'Oмская область': 'Омская область',
        'Hижегородская область': 'Нижегородская область',
        'г. Севастополь': 'г.Севастополь',
        'г.Москва': 'г.Москва',
        'г.Санкт-Петербург': 'г.Санкт-Петербург',
        'Чукотский авт.округ': 'Чукотский автономный округ',
    }
    df_clean['Регион'] = df_clean['Регион'].replace(replacements)
    df_clean['Регион'] = df_clean['Регион'].str.strip()
    return df_clean


def prepare_exog_data(df_exog, region, years, exog_cols):
    """
    Подготовка экзогенных переменных для региона.
    """
    region_data = df_exog[df_exog['Регион'] == region].copy()
    region_data = region_data[region_data['Год'].isin(years)].sort_values('Год')
    
    if len(region_data) != len(years):
        return None
    
    for col in exog_cols:
        if col in region_data.columns:
            region_data[col] = clean_numeric(region_data[col])
    
    return region_data[exog_cols].values


def fit_arimax(endog, exog_train, exog_forecast, order_range=(3, 2, 3)):
    """
    Подбор лучшей ARIMAX модели по AIC.
    """
    best_forecast = np.nan
    best_aic = np.inf
    best_order = None
    
    p_max, d_max, q_max = order_range
    
    for p in range(0, p_max):
        for d in range(0, d_max):
            for q in range(0, q_max):
                try:
                    model = ARIMA(
                        endog,
                        exog=exog_train,
                        order=(p, d, q)
                    )
                    fitted = model.fit()
                    forecast = fitted.forecast(steps=1, exog=exog_forecast)[0]
                    
                    if fitted.aic < best_aic and np.isfinite(forecast):
                        best_aic = fitted.aic
                        best_forecast = forecast
                        best_order = (p, d, q)
                except:
                    continue
    
    return best_forecast, best_aic, best_order


# ============================================================
# ЗАГРУЗКА ДАННЫХ
# ============================================================
print("Загрузка данных...")
df = pd.read_excel(FILE_PATH)

# Загрузка экзогенных данных
try:
    df_exog = pd.read_excel(EXOG_FILE)
    df_exog = standardize_region_names(df_exog)
    df_exog['Год'] = pd.to_numeric(df_exog['Год'], errors='coerce')
    USE_EXOG = True
    print(f"Экзогенные переменные загружены: {EXOG_COLUMNS}")
except FileNotFoundError:
    print(f"ВНИМАНИЕ: Файл {EXOG_FILE} не найден. Модель будет работать как обычная ARIMA.")
    USE_EXOG = False
    df_exog = None

# Стандартизация и очистка
df = standardize_region_names(df)
df['ОПЖ'] = clean_numeric(df['ОПЖ'])
df['Год'] = pd.to_numeric(df['Год'], errors='coerce').astype(int)
df = df[(df['Год'] >= 2014) & (df['Год'] <= 2023)]

# Группируем по регионам
regions = df['Регион'].unique()
results = []


# ============================================================
# ПРОГНОЗИРОВАНИЕ ARIMAX (валидация на 2023)
# ============================================================
for region in sorted(regions):
    ts = df[df['Регион'] == region].sort_values('Год')
    if len(ts) < 10:
        print(f"{region}: мало данных ({len(ts)} лет)")
        continue

    opzh_vals = ts['ОПЖ'].values
    years = ts['Год'].values
    
    train = opzh_vals[:-1]  # 2014-2022
    train_years = years[:-1]
    actual_2023 = opzh_vals[-1]
    test_year = [years[-1]]  # 2023

    # Подготовка экзогенных данных
    exog_train = None
    exog_forecast = None
    
    if USE_EXOG and df_exog is not None:
        exog_train = prepare_exog_data(df_exog, region, train_years, EXOG_COLUMNS)
        exog_forecast = prepare_exog_data(df_exog, region, test_year, EXOG_COLUMNS)
        
        if exog_train is None or exog_forecast is None:
            print(f"ВНИМАНИЕ: Нет экзогенных данных для {region}, используется ARIMA")
            exog_train = None
            exog_forecast = None

    # Подбор лучшей модели
    best_forecast, best_aic, best_order = fit_arimax(
        train, 
        exog_train, 
        exog_forecast
    )

    # Fallback
    if pd.isna(best_forecast):
        try:
            model = ARIMA(train, exog=exog_train, order=(1,1,1)).fit()
            best_forecast = model.forecast(steps=1, exog=exog_forecast)[0]
            best_order = (1, 1, 1)
        except:
            try:
                model = ARIMA(train, order=(1,1,1)).fit()
                best_forecast = model.forecast(steps=1)[0]
                best_order = (1, 1, 1)
            except:
                print(f"Не удалось для {region}")
                continue

    # Метрики
    rmse = np.sqrt(mean_squared_error([actual_2023], [best_forecast]))
    mae = mean_absolute_error([actual_2023], [best_forecast])
    
    model_type = "ARIMAX" if (exog_train is not None) else "ARIMA"

    results.append({
        'Регион': region,
        'Модель': model_type,
        'Порядок_(p,d,q)': str(best_order),
        'Прогноз_ОПЖ_2023': round(best_forecast, 4),
        'Факт_ОПЖ_2023': round(actual_2023, 4),
        'Абсолютная_ошибка': round(abs(best_forecast - actual_2023), 4),
        'RMSE': round(rmse, 4),
        'MAE': round(mae, 4)
    })

    print(f"{region} [{model_type}{best_order}]: прогноз = {best_forecast:.4f}, факт = {actual_2023:.4f}, ошибка = {abs(best_forecast - actual_2023):.4f}")


# ============================================================
# ИТОГИ
# ============================================================
if results:
    results_df = pd.DataFrame(results)
    overall_rmse = np.sqrt(mean_squared_error(results_df['Факт_ОПЖ_2023'], results_df['Прогноз_ОПЖ_2023']))
    overall_mae = mean_absolute_error(results_df['Факт_ОПЖ_2023'], results_df['Прогноз_ОПЖ_2023'])

    print("\n" + "="*60)
    print(f"Общий RMSE по ОПЖ: {overall_rmse:.4f}")
    print(f"Общий MAE по ОПЖ: {overall_mae:.4f}")
    print(f"Использованы экзогенные переменные: {USE_EXOG}")
    if USE_EXOG:
        print(f"Переменные: {EXOG_COLUMNS}")
    print("="*60)

    # Сохраняем результат
    output_file = "валидация_ОПЖ_прогноз_2023_ARIMAX.xlsx"
    results_df.to_excel(output_file, index=False)
    print(f"\nРезультаты сохранены в {output_file}")

    # Для Google Colab: скачивание файла
    try:
        from google.colab import files
        files.download(output_file)
    except:
        pass
else:
    print("Нет данных для обработки.")
```

# Пример формата файла с экзогенными переменными

Файл `exog_data.xlsx` должен иметь следующую структуру:

| Регион | Год | ВРП | Безработица | Расходы_на_здравоохранение | Доходы_населения |
|--------|-----|-----|-------------|---------------------------|------------------|
| Москва | 2014 | 12500 | 1.2 | 450000 | 65000 |
| Москва | 2015 | 13100 | 1.3 | 470000 | 68000 |
| ... | ... | ... | ... | ... | ... |
| Москва | 2024 | 18000 | 0.9 | 650000 | 95000 |

**Важно:**
- Столбцы `Регион` и `Год` обязательны
- Названия экзогенных столбцов должны совпадать с теми, что указаны в `EXOG_COLUMNS`
- Для прогноза на 2024 год нужны экзогенные данные за 2024 год (можно использовать прогнозные значения)
- Если экзогенные данные отсутствуют для какого-то региона, модель автоматически переключится на обычную ARIMA
